````python
import random
import faker
import faker_commerce
from mimesis import Generic
from datetime import datetime, timedelta, timezone
from typing import cast
from pprint import pprint

fake = faker.Faker("es_MX")
fake.add_provider(faker_commerce.Provider)
mim_generic = Generic(locale="es_MX")

#######################

productos = []
for _ in range(random.randint(50, 100)):
    producto = {
        "producto": cast(faker_commerce.Provider, fake.unique).ecommerce_name(),
        "categoria": cast(faker_commerce.Provider, fake).ecommerce_category(),
        "precio": cast(faker_commerce.Provider, fake).ecommerce_price(as_int=False)
        / 5000.0,
    }
    productos.append(producto)

database_name = "laboratorio"
collection_name = "productos"
mongo_client[database_name].drop_collection(collection_name)
mongo_client[database_name][collection_name].insert_many(productos)
print(f'{len(productos)} documentos insertados en colección "{collection_name}"')

#######################

clientes = []
for _ in range(random.randint(500, 1000)):
    client = {
        "nombre": cast(faker.Faker, fake.unique).name(),
        "edad": random.randint(18, 65),
        "contacto": {
            "telefono": cast(faker.Faker, fake.unique).phone_number(),
            "email": cast(faker.Faker, fake.unique).email(),
        },
        "compras": [],
    }
    for _ in range(random.randint(1, 5)):
        producto = random.choice(productos)
        cantidad = random.randint(1, 3)
        compra = {
            "producto_id": producto["_id"],
            "fecha": fake.past_datetime(
                start_date=datetime.now(tz=timezone.utc) - timedelta(days=365)
            ),
            "cantidad": cantidad,
            "monto": producto["precio"] * cantidad,
        }
        client["compras"].append(compra)
    clientes.append(client)

database_name = "laboratorio"
collection_name = "clientes"
mongo_client[database_name].drop_collection(collection_name)
mongo_client[database_name][collection_name].insert_many(clientes)
print(f'{len(clientes)} documentos insertados en colección "{collection_name}"')
````

Clasifica los clientes en buenos o malos.

- En una variable con nombre "buenos_clientes" almacena el resultado de una consulta que devuelva todos los clientes con al menos una compra de monto mayor o igual a 500

- En una variable con nombre "malos_clientes" almacena el resultado de una consulta que devuelva todos los clientes cuyas compras sean todas menores a 500

***En ambos casos devuelve el documento completo (es decir, sin omitir o agregar atributos, sin proyecciones, sin agrupaciones)***

````python
clientes_lab = mongo_client["laboratorio"]["clientes"] #lo separamos en una variable para no tener que escribir toda la ruta cada vez

buenos_clientes = list(
    clientes_lab.find(
        {"compras.monto": {"$gte": 500}})
)
malos_clientes = list(
    clientes_lab.find({
        "compras": {
            "$not": {
                "$elemMatch": {"monto": {"$gte": 500}}
            }
        }
    })
)
````

````python
print("Validando buenos_clientes...")
for cliente in buenos_clientes:
    print(cliente)
````

````python
print("Validando malos_clientes...")
for cliente in malos_clientes:
    print(cliente)
````

````python
# Validar intersección de clientes. Resultado esperado: 0 clientes en ambos grupos, ya que la clasificación es mutuamente excluyente.
ids_buenos = {cliente["_id"] for cliente in buenos_clientes}
ids_malos = {cliente["_id"] for cliente in malos_clientes}

interseccion = ids_buenos & ids_malos

print(f"Clientes en ambos grupos: {len(interseccion)}")

# Validar que la suma de clientes clasificados sea igual al total de clientes en la base de datos.
total_clientes = clientes_lab.count_documents({})

clasificados = len(ids_buenos | ids_malos) 

print(f"Total en BD: {total_clientes}")
print(f"Clasificados: {clasificados}")

# Validar que no haya clientes sin compras, ya que la clasificación se basa en el monto de las compras.
sin_compras = list(
    clientes_lab.find({"compras": {"$size": 0}})
)
print(f"Clientes sin compras: {len(sin_compras)}")
````

Calcula el total gastado por cada cliente
En una variable con nombre "gastos_totales" almacena la lista de los totales gastados por cada cliente con formato
```json
{
    "_id": ...,         // id del documento
    "nombre": ...,      
    "gastoTotal": ...
}
```

````python
clientes_lab = mongo_client["laboratorio"]["clientes"]
gastos_totales = list(
    clientes_lab.aggregate([ #aggregate se usa cuando quieres hacer mas operaciones como sumar, contar, etc. y el find es para consultas simples
        {"$project": { #$project elegimos las columnas que queremos devolver y agregamos columnas como queramos que se llamen, en este caso queremos el _id, el nombre y una columna nueva que se llama gastoTotal que es la suma de los montos de las compras
        #$project es como un select en SQL, pero también te permite crear nuevas columnas a partir de las existentes, como en este caso que creamos gastoTotal a partir de la suma de los montos de las compras
        #no estamos filtrando nada, por eso no hay un $match antes del $project, pero si quisieramos filtrar por algo, entonces si tendriamos que poner un $match antes del $project
        #no estamos creando nuevas columnas dentro de la colección, solo estamos creando nuevas columnas en el resultado de la consulta, por eso no usamos $addFields, sino que usamos $project
            "_id": 1, #esto es medio innecesario pq por dafault ya aparece el _id
            "nombre": 1,
            "gastoTotal": {"$sum": "$compras.monto"}
        }}
    ])
)
````

````python
print("Validando gastos_totales...")
print(gastos_totales)
````

Crea una consulta que devuelve todos los clientes enriqueciendo la información de sus compras con los datos de la collecion producto.

Es decir, cada resultado de la consulta debe seguir el esquema

```json
{
    "_id": ...,
    "compras": [
        {
            "_id": ...,
            "cantidad": ...,
            "categoria": ...,
            "fecha": ...,
            "monto": ...,
            "precio": ...,
            "producto": ...,
            "producto_id": ...
        },
        ...
    ],
    "contacto": {
        "email": ...,
        "telefono": ...
    },
    "edad": ...,
    "nombre": ...
}
```

***Almacena los resultados de la consulta en una variable con nombre "compras_enriquecidas"***

````python
clientes_lab = mongo_client["laboratorio"]["clientes"]

compras_enriquecidas = list(
    clientes_lab.aggregate([ #pipeline
        # 1. Descompone el array "compras". 
        #   Cada documento tendrá ahora UNA sola compra
        {"$unwind": "$compras"},

        # 2. Hace un join con la colección "productos"
        #   Relaciona compras.producto_id con productos._id
        #   El resultado se guarda en "producto_info" (como array)
        {"$lookup": {
            "from": "productos",
            "localField": "compras.producto_id",
            "foreignField": "_id",
            "as": "producto_info"
        }},

        # 3. Convierte "producto_info" de array a objeto
        #   (esperamos solo un producto por compra)
        {"$unwind": "$producto_info"},

        # 4. Enriquece el objeto "compras"
        #   Añade datos del producto directamente dentro de cada compra
        {"$addFields": {
            "compras.producto": "$producto_info.producto",
            "compras.categoria": "$producto_info.categoria",
            "compras.precio": "$producto_info.precio"
        }},

        # 5. Agrupa nuevamente por cliente (_id)
        #   Reconstruye el documento original con todas sus compras
        {"$group": {
            "_id": "$_id",
            "nombre": {"$first": "$nombre"},        # recupera nombre del cliente
            "edad": {"$first": "$edad"},            # recupera edad
            "contacto": {"$first": "$contacto"},    # recupera contacto
            "compras": {"$push": "$compras"}        # reconstruye el array de compras
        }},

        # 6. Proyección final
        #   Selecciona únicamente los campos deseados en la salida y su orden
        {"$project": {
            "_id": 1,
            "compras": 1,
            "contacto": 1,
            "edad": 1,
            "nombre": 1
        }}
    ])
)
````

````python
print("Validando compras_enriquecidas...")
# Imprimimos los primeros 5 documentos para revisar
from pprint import pprint

print(f"Se encontraron {len(compras_enriquecidas)} clientes con compras enriquecidas\n")
for cliente in compras_enriquecidas[:5]:
    pprint(cliente)
    print("-" * 80)
````
